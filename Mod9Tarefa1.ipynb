# Tarefa 1

### 1. Idade media das mães e dos pais por município (coluna munResNome)


```python
# 1. Agrupar por município de residência (munResNome)
# 2. Selecionar as colunas de idade (IDADEMAE, IDADEPAI)
# 3. Calcular a média (.mean()) para cada grupo

idade_media_por_municipio = sinasc_raw.groupby('munResNome')[['IDADEMAE', 'IDADEPAI']].mean()

print("Idade Média das Mães e Pais por Município de Residência:")
print(idade_media_por_municipio.round(2))
```

### 2. Peso médio dos bebes por sexo que nasceram no dia do seu aniversário por faixas de escolaridade mae
Ex: Você, aluna(o), nasceu no dia 10/01, então você precisa filtrar o conjunto de dados nessa data e calcular o peso médio dos bebês de cada sexo por faixa de escolaridade da mãe.

```python
sinasc_raw['DTNASC'] = pd.to_datetime(sinasc_raw['DTNASC'])

# Data de Aniversário: 06 de Agosto
dia_aniversario = 6
mes_aniversario = 8

# 1. Filtrar os nascimentos que ocorreram em 06 de Agosto
nascimentos_aniversario = sinasc_raw[
    (sinasc_raw['DTNASC'].dt.day == dia_aniversario) &
    (sinasc_raw['DTNASC'].dt.month == mes_aniversario)
]

# 2. Calcular o peso médio dos bebês (coluna PESO), agrupando por ESCMAE e SEXO
peso_medio_por_escolaridade_sexo = nascimentos_aniversario.groupby(
    ['ESCMAE', 'SEXO']
)['PESO'].mean().round(2)

# Exibir o resultado
print(f"Peso Médio dos Bebês (em {dia_aniversario:02d}/{mes_aniversario:02d}) por Escolaridade da Mãe e Sexo:")
print(peso_medio_por_escolaridade_sexo)
```

### 3. Qual o municipio que nasceu menos bebe em 2019?
    - qual a idade media, maxima, minima das maes nesse municipio?
    - qual a idade media, maxima, minima dos pais nesse municipio?

```python
# 1. Contar o número de nascimentos por município de residência
contagem_por_municipio = sinasc_raw['munResNome'].value_counts()

# 2. Encontrar o município com o menor número de nascimentos
municipio_min = contagem_por_municipio.idxmin()
nascimentos_min = contagem_por_municipio.min()

print(f"O município com menos nascimentos em 2019 foi: **{municipio_min}** (Total: {nascimentos_min} nascimentos)")
print("-" * 60)

# 3. Filtrar os dados para o município encontrado
dados_municipio_min = sinasc_raw[sinasc_raw['munResNome'] == municipio_min]

# 4. Calcular estatísticas de idade das mães e dos pais
estatisticas_idade = dados_municipio_min[['IDADEMAE', 'IDADEPAI']].agg(['mean', 'max', 'min']).round(2)

# 5. Exibir os resultados
print("\nEstatísticas de Idade das Mães neste Município:")
print(f"Média: {estatisticas_idade.loc['mean', 'IDADEMAE']} anos")
print(f"Máxima: {estatisticas_idade.loc['max', 'IDADEMAE']} anos")
print(f"Mínima: {estatisticas_idade.loc['min', 'IDADEMAE']} anos")

print("\nEstatísticas de Idade dos Pais neste Município:")
print(f"Média: {estatisticas_idade.loc['mean', 'IDADEPAI']} anos")
print(f"Máxima: {estatisticas_idade.loc['max', 'IDADEPAI']} anos")
print(f"Mínima: {estatisticas_idade.loc['min', 'IDADEPAI']} anos")

print("\n--- Tabela Completa ---")
print(estatisticas_idade)
```

### 4. Qual o municipio que nasceu mais bebe no mês de março?
    - qual a quantidade de filhos vivos media, maxima, minima nesse municipio?
    - qual a idade media, maxima, minima dos pais nesse municipio?



```python
# Mês de interesse: Março (mês 3)
mes_interesse = 3

# 1. Filtrar os nascimentos que ocorreram no mês de Março
nascimentos_marco = sinasc_raw[sinasc_raw['DTNASC'].dt.month == mes_interesse].copy()

# 2. Contar o número de nascimentos por município em Março
contagem_marco = nascimentos_marco['munResNome'].value_counts()

# 3. Encontrar o município com o maior número de nascimentos em Março
municipio_max_marco = contagem_marco.idxmax()
nascimentos_max_marco = contagem_marco.max()

print(f"O município com mais nascimentos no mês de Março foi: **{municipio_max_marco}** (Total: {nascimentos_max_marco} nascimentos)")
print("-" * 60)

# 4. Filtrar os dados para o município encontrado
dados_municipio_max_marco = sinasc_raw[sinasc_raw['munResNome'] == municipio_max_marco]

# 5. Calcular estatísticas de filhos vivos (QTDFILVIVO) e idade do pai (IDADEPAI)
estatisticas_marco = dados_municipio_max_marco[['QTDFILVIVO', 'IDADEPAI']].agg(['mean', 'max', 'min']).round(2)

# 6. Exibir os resultados
print("\nEstatísticas de Filhos Vivos (QTDFILVIVO) neste Município:")
print(f"Média: {estatisticas_marco.loc['mean', 'QTDFILVIVO']} filhos vivos")
print(f"Máxima: {estatisticas_marco.loc['max', 'QTDFILVIVO']} filhos vivos")
print(f"Mínima: {estatisticas_marco.loc['min', 'QTDFILVIVO']} filhos vivos")

print("\nEstatísticas de Idade dos Pais (IDADEPAI) neste Município:")
print(f"Média: {estatisticas_marco.loc['mean', 'IDADEPAI']} anos")
print(f"Máxima: {estatisticas_marco.loc['max', 'IDADEPAI']} anos")
print(f"Mínima: {estatisticas_marco.loc['min', 'IDADEPAI']} anos")

print("\n--- Tabela Completa ---")
print(estatisticas_marco)
```

Análise e Insight dos Dados SINASC
Um dado que costuma ser muito relevante em análises demográficas é a idade materna no parto e como ela se relaciona com fatores socioeconômicos, como a escolaridade.

Insight Principal: O Atraso na Maternidade em Mães com Alta Escolaridade
Ao calcular e visualizar a idade média das mães por sua faixa de escolaridade, é possível observar uma correlação clara e forte:

Quanto maior o nível de escolaridade da mãe, maior é a sua idade média ao ter o filho. Isso sugere que mulheres em Rondônia (em 2019) com mais tempo dedicado aos estudos estão, em média, adiando a maternidade.

```python
caminho_completo = "C:/Users/NOTE/Desktop/Pratica 1/Profissão Cientista de Dados M9 A1 - SINASC_RO_2019 (1).csv"
sinasc_raw = pd.read_csv(caminho_completo)

# Pré-processamento de Data e Cálculo da Média
sinasc_raw['DTNASC'] = pd.to_datetime(sinasc_raw['DTNASC'])
idade_por_escolaridade = sinasc_raw.groupby('ESCMAE')['IDADEMAE'].mean().sort_values()

# Exibir os resultados em tabela
print("Idade Média da Mãe por Escolaridade (em anos):")
print(idade_por_escolaridade.round(2))
print("-" * 50)

# Geração do Gráfico (com a correção do FutureWarning)
plt.figure(figsize=(10, 6))

sns.barplot(
    x=idade_por_escolaridade.index,
    y=idade_por_escolaridade.values,
    hue=idade_por_escolaridade.index, # Adicionado para corrigir o FutureWarning
    palette="viridis",
    legend=False
)

# Adiciona os valores nas barras
for index, value in enumerate(idade_por_escolaridade.values):
    plt.text(index, value + 0.1, f'{value:.2f}', ha='center', va='bottom', fontsize=10, weight='bold')

plt.title('Idade Média da Mãe por Escolaridade (SINASC - RO 2019)', fontsize=14)
plt.xlabel('Escolaridade da Mãe (Anos de Estudo)', fontsize=12)
plt.ylabel('Idade Média da Mãe (Anos)', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()

# O comando final que garante a exibição do gráfico no Jupyter
plt.show()
```
